```python
from collections import deque
```

```python
def bfs(graph, start, goal):
    visited = {start}
    queue = deque([start])
    parent = {start: None}
    order = []

    while queue:
        node = queue.popleft()
        order.append(node)

        if node == goal:
            path = []
            while node is not None:
                path.append(node)
                node = parent[node]
            path.reverse()
            return order, path

        for neighbour in graph.get(node, []):
            if neighbour not in visited:
                visited.add(neighbour)
                parent[neighbour] = node
                queue.append(neighbour)

    return order, None   # goal not reachable
```

```python
if __name__ == '__main__':
    # Maze represented as a graph (adjacency list)
    maze_graph = {
        'A': ['B', 'C'],
        'B': ['A', 'D', 'E'],
        'C': ['A', 'F'],
        'D': ['B'],
        'E': ['B', 'G'],
        'F': ['C', 'H'],
        'G': ['E'],
        'H': ['F']
    }
```

```python
start, goal = 'A', 'H'
order, path = bfs(maze_graph, start, goal)

print('Order of Node Expansion:', ' -> '.join(order))
if path:
        print('Shortest Path:', ' -> '.join(path))
        print('Path Length (edges):', len(path) - 1)
else:
        print(f'No path exists between {start} and {goal}')
```
